```python
# 51 participantes
# 3 sessões 
# 2 classes
# 59 canais
# 100 tiragens
import mne
import plotly.graph_objects as go
from IPython.display import display

from moabb.datasets import Yang2025

```

```python
# informações de um particiapnte

participant = 1
dataset = Yang2025()
data = dataset.get_data(subjects=[participant])

session_key = list(data[participant].keys())[0]
run_key = list(data[participant][session_key].key())[0]
raw = data[participant][session_key][run_key]

events, event_id = mne.events_from_annotations(raw)
epochs = mne.Epochs(raw, events, event_id=event_id,
tmin=-0.2, tmax=1.0, baseline=None, preload=True)

classes = list(event_id.keys())


dropdown_classe = widgets.Dropdown(
    options=classes,
    description="Classe:",
    style={'description_width': 'initial'}
)

slider_indice = widgets.IntSlider(
    value=0, min=0, max=len(imagens_por_classe[classes[0]]) - 1,
    description="Trial #",
    style={'description_width': 'initial'}
)

output = widgets.Output()

def mostrar_eeg(*args):
    with output:
        output.clear_output(wait=True)
        classe = dropdown_classe.value
        idx = slider_indice.value

        ep_classe = epochs[classe]   
        idx = min(idx, len(ep_classe) - 1)  
        trial = ep_classe[idx].get_data()[0]
        times = ep_classe.times

        fig = go.Figure()
        for ch_idx, ch_name in enumerate(epochs.ch_names):
            fig.add_trace(go.Scatter(
                x=times, y=trial[ch_idx],
                mode='lines', name=ch_name
            ))

        fig.update_layout(
            title=f"Sujeito {participant} | Classe: {classe} | Trial {idx+1}/{len(ep_classe)}",
            xaxis_title="Tempo (s)",
            yaxis_title="Amplitude (µV)",
            margin=dict(l=10, r=10, t=40, b=10),
            legend_title="Canal"
        )
        fig.show()

def ao_trocar_classe(change):
    nova_classe = change['new']
    max_idx = len(epochs[nova_classe]) - 1
    slider_indice.max = max_idx
    slider_indice.value = 0
    mostrar_eeg()

dropdown_classe.observe(ao_trocar_classe, names='value')
slider_indice.observe(mostrar_eeg, names='value')

display(widgets.VBox([dropdown_classe, slider_indice, output]))
mostrar_eeg()
```
